```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = pd.read_csv("tech_layoffs3.csv")
```

```python
df["reported_date"] = pd.to_datetime(df["reported_date"])
```

### Asking Questions as Data Analysis

### Q1) Which year laid off the most?

```python
df["reported_year"] = df["reported_date"].dt.year
```

```python
df.groupby("reported_year")["total_layoffs"].sum()
```

### Based on observation,2022 had the highest layoffs

### Q2) Which company had the highest layoff

```python
df2 = df.groupby("company")["total_layoffs"].sum().reset_index()
```

```python
df2.loc[df2["total_layoffs"] == df2["total_layoffs"].max()]
```

### Based on observation,TruePill had the highest layoffs

### Q3) Which industry had the highest layoff

```python
df.groupby("industry")["total_layoffs"].sum().sort_values(ascending = False)
```

### Healthcare had the highest layoff

```python
df.head(10)
```

```python
yearly_layoffs = (
    df.groupby(df["reported_year"])["total_layoffs"]
    .sum()
)

yearly_layoffs.plot(kind="bar")

plt.title("Yearly Layoffs")
plt.ylabel("Total Layoffs")
plt.show()
```

```python
top_companies = (
    df.groupby("company")["total_layoffs"]
    .sum()
    .sort_values(ascending=False)
    .head(10)
)

top_companies.plot(kind="bar")

plt.title("Top Companies by Layoffs")
plt.ylabel("Total Layoffs")
plt.show()
```

```python
industry_layoffs = (
    df.groupby("industry")["total_layoffs"]
    .sum()
    .sort_values(ascending=False)
    .head(10)
)

industry_layoffs.plot(kind="bar")

plt.title("Layoffs by Industry")
plt.ylabel("Total Layoffs")
plt.show()
```

```python
df["reported_year"] = pd.to_datetime(df["reported_year"])
```

### Key Insights

- 2022 had the highest number of layoffs.
- The layoffs trend declined after the peak period.
- Large tech companies contributed significantly to workforce reductions.
- Layoffs were concentrated in a few countries and industries.
- Post-IPO companies experienced substantial layoffs.
